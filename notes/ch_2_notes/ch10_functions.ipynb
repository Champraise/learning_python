# Functions
A function can take an input parameters and return a result. The result can be *Nothing* or *Something* a single object. 
* Actions on a function are Define it, with zero or more parameters
* Call it with zero or more arguments and get a result

```python
def do_nothing():
    pass

do_nothing()
```

```python
def make_a_sound():
    print("quack")

make_a_sound()
```

```python
def agree():
    return True
```

```python
if agree():
    print("Splendid!")
else:
    print("That was unexpected.")
```

```python
# Echo function - 'anything' is a parameter that can be any data type
def echo(anything):
    return anything + " " + anything
```

```python
# 'Rumplestiltskin' is an argument that is passed to the echo function
echo("Rumplestiltskin")
```

```python
def commentary(color):
    """Return a short description for a given color."""
    if color == "red":
        return "It's a tomato."
    elif color == "green":
        return "It's a green pepper."
    elif color == "bee purple":
        return "I don't know what it is, but only bees can see it."
    else:
        return "I've never heard of the color " + color + "."
```

```python
comment = commentary("bee purple")
print(comment)
```

```python
def whatis(thing):
    if thing is None:
        print("None")
    elif thing:
        print("True")
    else:
        print("False")
```

```python
whatis(None)
```

```python
whatis(True)
```

```python
whatis(False)
```

```python
whatis(0)
```

```python
whatis([])
```

```python
whatis(0.0)
```

### Positional Arguments

```python
def menu(wine, entree, dessert):
    return {"wine": wine, "entree": entree, "dessert": dessert}
```

```python
menu('chardonnay', 'chicken', 'cake')
```

```python
def buggy(arg, result=[]):
    result.append(arg)
    print(result)
```

```python
buggy('a')
```

```python
buggy('a')
```

```python
def works(arg):
    result = []
    result.append(arg)
    print(result)
```

```python
works('a')
```

```python
works('b')
```

```python
def print_args(*args):
    """Print any number of arguments."""
    print('Positional argument tuple:', args)
```

```python
print_args(1, 2, 3)
```

```python
def print_more(required1, required2, *args):
    print('Need this one:', required1)
    print('Need this one too:', required2)
    print('All the rest:', args)
```

```python
print_more('cap', 'gloves', 'scarf', 'monocle', 'mustache wax')
```

```python
def print_kwargs(**kwargs):
    """Print any number of keyword arguments."""
    print('Keyword argument dictionary:', kwargs)
```

```python
print_kwargs()
```

```python
print_kwargs(wine='merlot', entree='mutton', dessert='macaroon')
```

```python
def print_data(data, /, start=0, end=100):
    for value in (data[start:end]):
        print(value)
```

```python
def outer(a, b):
    def inner(c, d):
        return c + d
    return inner(a, b)
```

```python
outer(4, 7)
```

```python
def edit_story(words, func):
    for word in words:
        print(func(word))
```

```python
stairs = ["thud", "meow", "thud", "hiss"]
```

```python
def enliven(word):
    return word.capitalize() + "!"
```

```python
edit_story(stairs, enliven)
```

```python
edit_story(stairs, lambda word: word.capitalize() + "!")
```

```python
def my_range(first=0, last=10, step=1):
    number = first
    while number < last:
        yield number
        number += step
```

```python
my_range
```

```python
ranger = my_range(1, 5)
```
