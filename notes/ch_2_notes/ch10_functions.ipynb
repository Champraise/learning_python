# Functions
A function can take an input parameters and return a result. The result can be *Nothing* or *Something* a single object. 
* Actions on a function are Define it, with zero or more parameters
* Call it with zero or more arguments and get a result

```python
def do_nothing():
    pass

do_nothing()
```

```python
def make_a_sound():
    print("quack")

make_a_sound()
```

```python
def agree():
    return True
```

```python
if agree():
    print("Splendid!")
else:
    print("That was unexpected.")
```

```python
# Echo function - 'anything' is a parameter that can be any data type
def echo(anything):
    return anything + " " + anything
```

```python
# 'Rumplestiltskin' is an argument that is passed to the echo function
echo("Rumplestiltskin")
```

```python
def commentary(color):
    """Return a short description for a given color."""
    if color == "red":
        return "It's a tomato."
    elif color == "green":
        return "It's a green pepper."
    elif color == "bee purple":
        return "I don't know what it is, but only bees can see it."
    else:
        return "I've never heard of the color " + color + "."
```

```python
comment = commentary("bee purple")
print(comment)
```
