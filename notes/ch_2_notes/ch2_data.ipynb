# Introducing Python Objects

```python
# Creating a bytes object
my_bytes = b"hello" 

# If we try to change the first letter 'h' (which is the number 104) to 'J' (74)
my_bytes[0] = 74  

# This will immediately throw a TypeError! You cannot edit it.
```

```python
# Creating a bytearray
my_bytearray = bytearray(b"hello")

# Let's change the first letter 'h' (104) to 'J' (74)
my_bytearray[0] = 74 

# This works perfectly!
print(my_bytearray) 
# Output: bytearray(b'Jello')
```
