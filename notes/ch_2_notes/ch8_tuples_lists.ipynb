```python
# iterate multiple sequences with zip()
days = ['Monday', 'Tuesday', 'Wednesday']
activities = ['Gym', 'Class', 'Study']
drinks = ['Water', 'Soda', 'Juice']
for day, activity, drink in zip(days, activities, drinks):
    print(f"On {day} I have {activity} and drink {drink}")
```
