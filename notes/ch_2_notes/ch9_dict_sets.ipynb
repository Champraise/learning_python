```python
signals = {
    'green': "go",
    'yellow': "go faster",
    'red': "smile for the camera"
}
```

```python
save_signals = signals
signals['blue'] = 'confuse everyone'
print(save_signals)
```

```python
# to avoid this, use the copy method
signals = {
    'green': "go",
    'yellow': "go faster",
    'red': "smile for the camera"
}
save_signals = signals.copy()
signals['blue'] = 'confuse everyone'
signals
```

```python
# this a shallow copy if you need a deep copy for mutable dict values use deepcopy from the copy module
save_signals
```

```python
word = 'letters'
letter_counts = {letter: word.count(letter) for letter in set(word)}
letter_counts
```

```python
vowels = 'aeiou'
word = 'onomatopoeia'
vowel_counts = {letter: word.count(letter) for letter in set(word) if letter in vowels}
vowel_counts
```

```python
drinks = {
    'martini': {'vodka', 'vermouth'},
    'black russian': {'vodka', 'kahlua'},
    'white russian': {'vodka', 'kahlua', 'cream'},
    'manhattan': {'rye', 'vermouth', 'bitters'},
    'screwdriver': {'vodka', 'orange juice'}
}
```

```python
for name, contents in drinks.items():
    if 'vodka' in contents:
        print(name, contents)
```

```python
for name, contents in drinks.items():
    if 'vodka' in contents and not ('vermouth' in contents or 'cream' in contents):
        print(name, contents)
```

```python
for name, contents in drinks.items():
    if contents & {'vermouth', 'cream'}:
        print(name)
```

```python
for name, contents in drinks.items():
    if 'vodka' in contents and not contents & {'vermouth', 'cream'}:
        print(name)
```

```python
bruss = drinks['black russian']
wruss = drinks['white russian']
```
